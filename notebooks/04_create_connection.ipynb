```python
# Ejecutar después de 00_setup y de crear el Secret Scope electrocasa-sql.
# Exige CREATE CONNECTION y CREATE CATALOG en el metastore.
# El SQL no incrusta el usuario ni la contraseña: los resuelve Unity Catalog.
spark.sql("""CREATE CONNECTION IF NOT EXISTS electrocasa_sql TYPE sqlserver
OPTIONS (
  host 'analyticsdmc.database.windows.net',
  port '1433',
  user secret ('electrocasa-sql','username'),
  password secret ('electrocasa-sql','password')
)""")
spark.sql("""CREATE FOREIGN CATALOG IF NOT EXISTS electrocasa_sql
USING CONNECTION electrocasa_sql OPTIONS (database 'electrocasadb')""")
print('Catálogo foráneo creado. Comprueba SELECT en electrocasa_sql.dbo.TrackingEnvios')
```

```python
# Si falla, el problema es la conexión, permiso SELECT sobre el catálogo
# foráneo o la disponibilidad/red del servidor SQL.
display(spark.sql('SELECT tracking_id, estado_entrega FROM electrocasa_sql.dbo.TrackingEnvios LIMIT 5'))
```
